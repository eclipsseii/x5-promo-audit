# 03. Профиль клиента
Агрегация истории чеков (до 18.03.2019) в признаки клиента. Запрос: `sql/01_client_features.sql`.

```python
import duckdb
import pandas as pd

con = duckdb.connect()

sql = open('sql/01_client_features.sql').read()
con.execute(f"COPY ({sql}) TO 'data/features.csv' (HEADER)")

f = pd.read_csv('data/features.csv')
print(f.shape, '| клиентов без покупок:', (f['n_checks'] == 0).sum())
print(f.drop(columns='client_id').describe().T.round(2))
```

**Итог:**
- Профиль собран для всех 200 039 клиентов, у всех есть покупки.
- recency_days от 1 до 23 дней: в эксперимент попали только недавно активные клиенты.
- share_promo_checks ≈ 0 (экспресс-баллы почти не начислялись) → не используется. Склонность к выгоде оценивается по share_points_checks — доле чеков со списанием баллов.
